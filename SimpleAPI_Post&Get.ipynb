```python
import requests
```

```python
url_get = 'https://api.coingecko.com/api/v3/simple/price?ids=bitcoin&vs_currencies=usd'
response = requests.get(url_get)
status_code = response.status_code

print(response.json())
```

for API test in CMD:
curl -X GET "https://api.coingecko.com/api/v3/simple/price?ids=bitcoin&vs_currencies=usd"

```python
url_post = 'https://jsonplaceholder.typicode.com/posts'

data = {
    'title': 'foo',
    'body': 'bar'
}

response = requests.post(url_post,data)
print(response.status_code)
print(response.json())
```

API test in CMD: 
curl -X POST "https://jsonplaceholder.typicode.com/posts" -H "Content-Type: application/json" -d "{\"title\":\"foo\",\"body\":\"bar\",\"userId\":1}"
